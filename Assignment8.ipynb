# Q1. Text preprocessing and word frequency distribution

**Objective:** To convert text to lowercase, remove punctuation, tokenize words and sentences, remove stopwords, and display word frequency distribution.

The following paragraph is used for both Q1 and Q2.

```python
# Original paragraph
text = """Artificial intelligence is one of my favorite topics in technology.
It is transforming the world by making machines smarter and more useful.
AI helps people solve complex problems in healthcare, education, and business.
Machine learning allows computers to learn from data and improve their performance.
I enjoy learning about AI because it creates exciting opportunities for innovation.
In the future, artificial intelligence will play an important role in our daily lives."""

print("Original Text:\n")
print(text)

# 1. Convert text to lowercase and remove punctuation
clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("\nCleaned Text:\n")
print(clean_text)

# 2. Tokenize text into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:")
print(words)

print("\nSentence Tokens:")
print(sentences)

# 3. Compare split() and word_tokenize()
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("\nUsing split():")
print(split_words)

print("\nUsing word_tokenize():")
print(nltk_words)

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in nltk_words if word not in stop_words]

print("\nWords after Stopword Removal:")
print(filtered_words)

# 5. Word Frequency Distribution
freq = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, count in freq.most_common():
    print(f"{word} : {count}")
```

### Q1 – Explanation and observations

1. **Lowercase conversion:** `lower()` converts all characters to lowercase.
2. **Punctuation removal:** `re.sub()` removes punctuation from the text.
3. **Tokenization:** `word_tokenize()` divides text into word tokens, while `sent_tokenize()` divides it into sentences.
4. **Stopword removal:** Common words such as `is`, `the`, and `and` are removed using NLTK's English stopword list.
5. **Frequency distribution:** `FreqDist()` counts how often each remaining word occurs.

**Difference between `split()` and `word_tokenize()`:**

| `split()` | `word_tokenize()` |
|---|---|
| Built-in Python function | NLTK tokenization function |
| Splits primarily on whitespace | Uses NLP tokenization rules |
| Simple and fast | More suitable for NLP tasks |
| Does not apply language-aware tokenization rules | Handles punctuation and contractions more carefully |

Because punctuation was removed before this comparison, the two outputs may be very similar for this paragraph.

# Q2. Feature extraction, stemming and lemmatization

**Objective:** To extract alphabetic words, remove stopwords, perform stemming and lemmatization, and compare the results.

```python
# Same paragraph used in Q1
text = """Artificial intelligence is one of my favorite topics in technology.
It is transforming the world by making machines smarter and more useful.
AI helps people solve complex problems in healthcare, education, and business.
Machine learning allows computers to learn from data and improve their performance.
I enjoy learning about AI because it creates exciting opportunities for innovation.
In the future, artificial intelligence will play an important role in our daily lives."""

# 1. Extract words containing only alphabets
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Extracted Words:")
print(words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("\nAfter Stopword Removal:")
print(filtered_words)

# 3. Perform stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nStemmed Words:")
print(stemmed_words)

# 4. Perform lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Compare stemming and lemmatization
print("\nComparison:")
for original, stem, lemma in zip(filtered_words, stemmed_words, lemmatized_words):
    print(f"{original} -> Stem: {stem} | Lemma: {lemma}")
```

### Q2 – Explanation and comparison

- **Alphabetic word extraction:** `re.findall(r'\\b[a-zA-Z]+\\b', ...)` extracts sequences containing only English alphabetic characters.
- **Stopword removal:** Common English stopwords are removed before applying stemming and lemmatization.
- **Stemming:** `PorterStemmer` reduces words by applying rule-based suffix removal. The result may not be a valid dictionary word.
- **Lemmatization:** `WordNetLemmatizer` converts words to dictionary base forms. Its default part of speech is noun.

**Examples (the exact output depends on the input word):**

| Original | Stemmed | Lemmatized (noun default) |
|---|---|---|
| machines | machin | machine |
| computers | comput | computer |
| opportunities | opportun | opportunity |
| problems | problem | problem |

**When to prefer each method:**
- Prefer **stemming** when speed is important and a rough root form is sufficient, such as in some search or information-retrieval tasks.
- Prefer **lemmatization** when meaningful dictionary words are important, such as in text understanding and linguistic analysis.

**Conclusion:** Stemming is generally faster but can produce non-dictionary forms. Lemmatization is more linguistically meaningful, though it can require more processing. Supplying the correct part of speech can improve lemmatization accuracy.

# Q3. Feature extraction using Bag of Words and TF-IDF

**Objective:** To represent three short texts numerically using `CountVectorizer` and `TfidfVectorizer`, then identify the top three keywords in each text.

```python
# Three sample texts on different topics
documents = [
    "Artificial intelligence is transforming technology and improving automation.",
    "Football is a popular sport played by teams around the world.",
    "Healthy food provides energy and improves physical health."
]

# 1. Bag of Words using CountVectorizer
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(documents)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

# 2. TF-IDF using TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTF-IDF Vocabulary:")
print(feature_names)

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

# 3. Find the top 3 keywords from each text
print("\nTop 3 Keywords from Each Text:")

for i, document in enumerate(documents):
    scores = tfidf_matrix[i].toarray()[0]

    # Indices of the three highest TF-IDF scores
    top_indices = scores.argsort()[-3:][::-1]

    print(f"\nDocument {i + 1}: {document}")
    for index in top_indices:
        print(f"{feature_names[index]} : {scores[index]:.4f}")
```

### Q3 – Explanation and interpretation

**1. Bag of Words (`CountVectorizer`)**
- Creates a vocabulary of unique words from all documents.
- Represents each document as a vector of word counts.
- Does not consider the importance or meaning of a word.

**2. TF-IDF (`TfidfVectorizer`)**

TF-IDF stands for *Term Frequency–Inverse Document Frequency*. It assigns a weight to a word based on its frequency in a document and how uncommon it is across the collection of documents.

\[
TFIDF(t,d) = TF(t,d) \times IDF(t)
\]

- **TF (Term Frequency):** Measures how frequently a term appears in a document.
- **IDF (Inverse Document Frequency):** Gives less weight to terms that appear in many documents and more weight to terms that are relatively uncommon.

**3. Top three keywords**

The code sorts each document's TF-IDF scores in descending order and prints the three highest-scoring terms. Since each sample text covers a different topic, the top terms should help distinguish technology, sports, and food.

**Difference between Bag of Words and TF-IDF:**

| Bag of Words | TF-IDF |
|---|---|
| Counts word occurrences | Measures word importance |
| Uses raw term counts | Uses term frequency and inverse document frequency |
| Common words can have high counts | Words common across documents receive lower weights |
| Useful as a simple text representation | Useful in search and text classification |

**Conclusion:** CountVectorizer represents documents using word counts, while TF-IDF gives greater weight to terms that are more informative within a document collection.
