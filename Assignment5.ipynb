```python
#Question 4
Your_Name = np.linspace(10, 100, 25)

print("Array:\n", Your_Name)
print("Dimensions:", Your_Name.ndim)
print("Shape:", Your_Name.shape)
print("Total Elements:", Your_Name.size)
print("Data Type:", Your_Name.dtype)
print("Bytes Consumed:", Your_Name.nbytes)
transpose_arr = Your_Name.reshape(25,1)
print("\nTranspose using reshape():")
print(transpose_arr)
print("\nUsing T attribute:")
print(Your_Name.T)
print("\nNote: T has no effect on a 1-D array.")
```

```python
#Question 5
ucs420_your_name = np.array([
    [10,20,30,40],
    [50,60,70,80],
    [90,15,20,35]
])

print("Original Array:\n", ucs420_your_name)

print("Mean:", np.mean(ucs420_your_name))
print("Median:", np.median(ucs420_your_name))
print("Maximum:", np.max(ucs420_your_name))
print("Minimum:", np.min(ucs420_your_name))
print("Unique Elements:", np.unique(ucs420_your_name))

# Reshape
reshaped_ucs420_your_name = ucs420_your_name.reshape(4,3)

print("\nReshaped Array (4x3):")
print(reshaped_ucs420_your_name)

# Resize
resized_ucs420_your_name = np.resize(ucs420_your_name, (2,3))

print("\nResized Array (2x3):")
print(resized_ucs420_your_name)
```
