```python
import pandas as pd

# Fixed Entries
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Personalized Entries (Roll No: 1024170012)
personal_entries = [
    {
        "question": "How do I update my registered mobile number?",
        "answer": "Go to Profile > Edit Mobile Number.",
        "keywords": "mobile update profile",
        "category": "account"
    },
    {
        "question": "Where is the examination office?",
        "answer": "The examination office is located in Block A.",
        "keywords": "exam office location",
        "category": "general"
    }
]

faq_df = pd.DataFrame(fixed_entries + personal_entries)

print(faq_df)
```

```python
def score_query(query, df):

    query_words = query.lower().split()

    scores = []

    for index, row in df.iterrows():

        text = (row["question"] + " " + row["keywords"]).lower()

        score = 0

        for word in query_words:
            if word in text:
                score += 1

        if score > 0:
            scores.append((score, row))

    scores.sort(reverse=True, key=lambda x: x[0])

    return scores


query = input("Enter your query: ")

result = score_query(query, faq_df)

for score, row in result:
    print("\nScore:", score)
    print("Question:", row["question"])
    print("Answer:", row["answer"])
```

```python
def same_category(category_name, df):

    return df[df["category"] == category_name]


print("Questions in Account Category\n")

print(same_category("account", faq_df))
```

```python
def search_with_ties(query, df):

    query_words = query.lower().split()

    scores = []

    for index, row in df.iterrows():

        text = (row["question"] + " " + row["keywords"]).lower()

        score = 0

        for word in query_words:
            if word in text:
                score += 1

        scores.append(score)

    highest = max(scores)

    if highest == 0:
        print("No Match Found")
        return

    print("Highest Score =", highest)

    for i in range(len(scores)):
        if scores[i] == highest:
            print("\nQuestion:", df.iloc[i]["question"])
            print("Answer:", df.iloc[i]["answer"])
            print("Category:", df.iloc[i]["category"])


print("Tie Example")
search_with_ties("fee", faq_df)

print("\nNon-Tie Example")
search_with_ties("password", faq_df)
```
