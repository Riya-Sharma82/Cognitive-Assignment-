```python
# 1. Sum all items in a list

numbers = [10, 20, 30, 40, 50]
total = sum(numbers)

print("1. Sum:", total)


# 2. Get the largest number from a list

numbers = [10, 45, 23, 89, 12, 67]
largest = max(numbers)

print("2. Largest number:", largest)


# 3. Sort tuples by the last element

items = [(2, 5), (1, 2), (4, 4), (2, 3)]
result = sorted(items, key=lambda x: x[-1])

print("3. Sorted list:", result)


# 4. Remove duplicates from a list

numbers = [1, 2, 3, 2, 4, 1, 5, 3]
result = list(dict.fromkeys(numbers))

print("4. List after removing duplicates:", result)


# 5. Find words longer than n

words = ["apple", "banana", "cat", "elephant", "dog", "computer"]
n = 5

result = [word for word in words if len(word) > n]

print("5. Words longer than", n, ":", result)


# 6. Find difference between two lists

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

difference = list(set(list1) - set(list2))

print("6. Difference:", difference)


# 7. Add a key to a dictionary

student = {
    "name": "Riya",
    "age": 20
}

student["branch"] = "COPC"

print("7. Dictionary after adding key:", student)


# 8. Concatenate dictionaries

dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

result = {}

result.update(dic1)
result.update(dic2)
result.update(dic3)

print("8. Concatenated dictionary:", result)


# 9. Remove a key from a dictionary

student = {
    "name": "Riya",
    "age": 20,
    "branch": "COPC"
}

del student["age"]

print("9. Dictionary after removing key:", student)


# 10. Multiply all items in a dictionary

numbers = {
    "a": 2,
    "b": 3,
    "c": 4
}

result = 1

for value in numbers.values():
    result *= value

print("10. Product:", result)
```
