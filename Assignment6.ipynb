```python
##Question 1
import numpy as np
#(a)
temperature=np.array([25,28,31,35,38,27,33,40])
corrected_temp=temperature+2
print("The Original temperature",temperature)
print("The Corrected temperature",corrected_temp)
#(b)
fahrenheit=(9/5)*corrected_temp+32
print("The converted temperature in fahrenheit",fahrenheit)
#(c)
greater_than_32=corrected_temp[corrected_temp>32]
print("Reading greater than 32°C",greater_than_32)
#(d)
count=np.sum(corrected_temp>32)
print("Number of reading :",count)
```

```python
##Question 2
import numpy as np
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
#(a)
total_steps=np.sum(steps)
print("Total Steps:",total_steps)
#(b)
mean_steps=np.mean(steps)
print("Mean Steps:",mean_steps)
#(c)
maximum=np.max(steps)
minimum=np.min(steps)
print("The Maximum steps:",maximum)
print("The Minimum steps:",minimum)
#(d)
day_total=np.sum(steps,axis=0)
print("Total steps in for each day",day_total)
#(e)
user_total=np.sum(steps,axis=1)
print("Total steps in for each user",user_total)
```

```python
import numpy as np

# a. Create original array
original = np.array([1, 2, 3, 4, 5, 6])

print("Original array:", original)


# b. Slice from index 1 to 4
subset = original[1:5]

print("Subset:", subset)


# c. Modify first element of subset
subset[0] = 999

print("\nAfter modifying subset:")
print("Original:", original)
print("Subset:", subset)


# d. Create another slice using copy()
copied = original[1:5].copy()

copied[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied:", copied)


# e. Create numbers 1 to 12 and reshape into 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)

print("\n3 x 4 Matrix:")
print(matrix)


# f. Indexing and slicing

# First row
first_row = matrix[0]

# Last row
last_row = matrix[-1]

# Second column
second_column = matrix[:, 1]

# Rows 1-2 and columns 2-3
selected_elements = matrix[1:3, 2:4]

print("\nFirst row:", first_row)
print("Last row:", last_row)
print("Second column:", second_column)
print("Rows 1-2 and columns 2-3:")
print(selected_elements)


# g. Flatten and ravel
flattened = matrix.flatten()
raveled = matrix.ravel()

print("\nFlatten:", flattened)
print("Ravel:", raveled)


# h. Modify ravel array
raveled[0] = 1000

print("\nAfter modifying ravel:")
print("Ravel:", raveled)
print("Original matrix:")
print(matrix)


# i. Modify flatten array
flattened[1] = 2000

print("\nAfter modifying flatten:")
print("Flatten:", flattened)
print("Original matrix:")
print(matrix)


# j. Display shape, ndim, size and dtype
print("\nMatrix properties:")
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y=np.array([40, 65, 30, 85])
print("Shape of X",X.shape)
print("Dimension of X",X.ndim)
#(b)
X_t=X.T
print("\nTranspose of X:",X_t)
#(c)
XT_X = X.T @ X

print("\nX.T @ X:")
print(XT_X)
#(d)
inverse = np.linalg.inv(XT_X)

print("\nInverse of X.T @ X:")
print(inverse)
#(e)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("\nOLS coefficients:")
print(beta)

#(f)
print("\nCoefficients:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])
#(g)
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("\nPredicted assistance score:", predicted_score)
```
