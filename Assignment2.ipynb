```python
import random
from collections import Counter

# Part (i): Generate 100 random numbers
random.seed(1024170012)
nums = [random.randint(100, 900) for i in range(100)]

print("Random Numbers:")
print(nums)

# Part (ii): Find odd numbers
odd = [x for x in nums if x % 2 != 0]
print("Odd Numbers:", odd)
print("Count of Odd Numbers:", len(odd))

# Part (iii): Find even numbers
even = [x for x in nums if x % 2 == 0]
print("Even Numbers:", even)
print("Count of Even Numbers:", len(even))

# Part (iv): Find prime numbers
def prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

primes = [x for x in nums if prime(x)]
print("Prime Numbers:", primes)
print("Count of Prime Numbers:", len(primes))

# Part (v): Find most frequent number
c = Counter(nums)
num, count = c.most_common(1)[0]

print("Most Frequent Number:", num)
print("Occurrences:", count)
```

```python

roll = "10241700"

A = {int(d) * 7 for d in roll}
B = {int(d) * 9 for d in roll}

print("Set A:", A)
print("Set B:", B)

# (vi) Union
print("Union:", A.union(B))

# (vii) Intersection
print("Intersection:", A.intersection(B))

# (viii) Difference
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

print("Difference() gives elements in one set but not the other, while symmetric_difference() gives elements unique to both sets.")

# (ix) Symmetric Difference
print("Symmetric Difference:", A.symmetric_difference(B))

# (x) Subset and Superset
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))

# (xi) Remove a value using discard()
x = int(input("Enter value to remove from A: "))
A.discard(x)

print("Updated Set A:", A)

print("discard() is safer because it does not raise an error if the element is not present.")
```

```python
my_dict = {
    "name": "Riya Sharma",
    "roll_no": "1024170012",
    "branch": "CSE",
    "age": 19,
    "city": "Jaipur"
}

# (i) Rename city to location
my_dict["location"] = my_dict.pop("city")
print("After renaming city to location:")
print(my_dict)

# (ii) Add CGPA
my_dict["cgpa"] = 6.75
print("\nAfter adding CGPA:")
print(my_dict)

# (iii) Increase age by 1
my_dict["age"] += 1
print("\nAfter increasing age:")
print(my_dict)

# (iv) Delete branch using pop()
d1 = my_dict.copy()
removed = d1.pop("branch")
print("\nUsing pop():")
print(d1)
print("Removed value:", removed)

# Delete branch using del
d2 = my_dict.copy()
del d2["branch"]
print("\nUsing del:")
print(d2)

print("\npop() removes and returns the value, while del only removes the key.")

# (v) Print key-value pairs
print("\nKey-Value Pairs:")
for key, value in my_dict.items():
    print(key, "→", value)

# (vi) Check for email key
print("\nChecking for email:")
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key not found")

# (vii) Create friend dictionary and merge
friend_dict = {
    "name": "Aman",
    "roll_no": "1024170001",
    "branch": "IT",
    "age": 20,
    "city": "Delhi"
}

merged = {**my_dict, **friend_dict}

print("\nMerged Dictionary:")
print(merged)

print("When keys are the same, values from the second dictionary win.")

# (viii) Dictionary comprehension
string_dict = {k: v for k, v in my_dict.items() if type(v) == str}

print("\nDictionary with only string values:")
print(string_dict)
```
